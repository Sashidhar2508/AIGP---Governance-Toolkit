# AIGP Risk & Governance Walkthrough

Interactive notebook for the **AIGP Governance Toolkit** portfolio project.

Demonstrates:
- Loading a demo AI use case
- Mapping to NIST AI RMF, EU AI Act signals, and ISO/IEC 42001
- Running risk scoring, DPIA draft, and conformity-style checklist
- Interpreting results for Technical Account Manager / AI Governance discussions

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from src.frameworks import list_use_cases, full_framework_map
from src.risk_assessor import assess
from src.dpia import generate_dpia
from src.conformity import generate_conformity_checklist

print("Available use cases:")
for k in list_use_cases():
    print(f"  • {k}")
```

## 1. Select & Inspect a Use Case

```python
key = "enterprise_llm_support_agent"
fmap = full_framework_map(key)
uc = fmap["use_case"]

print(f"Name: {uc['name']}")
print(f"Description: {uc['description']}")
print(f"Data categories: {uc['data_categories']}")
print(f"Automated decision: {uc['automated_decision']}")
print(f"Affects individuals: {uc['affects_individuals']}")
print(f"High-stakes domain: {uc['high_stakes_domain']}")
print(f"Third-party model: {uc['third_party_model']}")
```

## 2. Risk Assessment

```python
risk = assess(uc)
print(f"Risk Score : {risk['risk_score']}")
print(f"Risk Level : {risk['risk_level']}")
print(f"DPIA Rec.  : {risk['dpia_recommended']}")
print(f"HITL Rec.  : {risk['hitl_recommended']}")
print("\nFlags:")
for f in risk["risk_flags"]:
    print(f"  • {f}")
print("\nKey Findings:")
for finding in risk["key_findings"]:
    print(f"  – {finding}")
```

## 3. Framework Mapping Snapshot

```python
print("EU AI Act tier hint:", fmap["eu_ai_act"]["tier_hint"])
print("\nNIST AI RMF – Govern activities (sample):")
for a in fmap["nist_ai_rmf"]["Govern"][:3]:
    print(f"  • {a}")
print("\nISO/IEC 42001 evidence points (sample):")
for a in fmap["iso_42001"][:4]:
    print(f"  • {a}")
```

## 4. DPIA & Conformity Drafts

```python
dpia = generate_dpia(uc, risk)
conf = generate_conformity_checklist(uc, risk)

print("DPIA Title:", dpia["title"])
print("DPIA required indicator:", dpia["dpia_required_indicator"])
print("\nRisks to rights & freedoms:")
for r in dpia["risks_to_rights_and_freedoms"]:
    print(f"  • {r}")

print("\nConformity checklist items:", len(conf["checklist"]))
print("Applicable under heuristics:", conf["applicable"])
```
